```python
def flight_dep_time(x):

    if (x>4) and (x<=8):
        return "Early Morning"
        
    elif (x>8) and (x<=12):
        return "Morning"

    elif (x>12) and (x<=16):
        return "Afternoon"

    elif (x>16) and (x<=20): 
        return "Evening"

    elif (x>20) and (x<=24):
        return "Night"

    else:
        return "Late Night"
    
```

```python
data["Dep_Time_hour"].apply(flight_dep_time).value_counts().plot(kind="bar", color='g')
```

```python
def preprocess_duration(x):

    if 'h' not in x:
        x = '0h'+ ' '+ x
    elif 'm' not in x:
        x = x + ' '+ '0m'
    return x    
```

```python
data["Duration"] = data["Duration"].apply(preprocess_duration)
```
